```python
from math import ceil
import matplotlib.pyplot as plt
import os
```

### the distance  and depths of sparse points for the `dome_decenter_x_short` dataset

```python
evaluation_path = "/home/hendrik/masterproject/evaluation_2025_01_23/dome_decenter_x_short"
cols = 2
num_images = 106

# for i in range(1, num_images+1):
for i in [1,2,3]:
    px = 1/plt.rcParams['figure.dpi']
    fig, axs = plt.subplots(1,2, figsize=(1200*px, 480*px))
    with open(os.path.join(evaluation_path, f"img_{i}_distances")) as distance_file:
        distances = list(map(float, distance_file.readlines()))
    with open(os.path.join(evaluation_path, f"img_{i}_depths")) as depths_file:
        depths = list(map(float, depths_file.readlines()))

    axs[0].hist(distances, bins=100)
    axs[0].set_title("Distances")
    axs[0].set_xlabel("distance in m")
    axs[0].set_ylabel("number of samples")
    axs[1].hist(depths, bins=100)
    axs[1].set_title("Depths")
    axs[1].set_xlabel("depth in m")
    axs[1].set_ylabel("number of samples")
    fig.suptitle(f"Image {i}")
    plt.show()
```
